```python
import numpy as np
import pandas as pd
```

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})

print(employees)
print("\n")
print(salaries)
```

```python
#mergin in basically like join of sql 
#has types like inner outer left right

print(pd.merge(employees, salaries, on = 'employee_id', how = 'inner'))
#if we dont mention anything employee id is the only common column in both
#also inner join is done by default

print(pd.merge(employees, salaries, ))
```

```python
df_left = pd.merge(employees, salaries, on='employee_id', how='left')

print(df_left)

# SQL Equivalent:
# SELECT *
# FROM employees
# LEFT JOIN salaries
# ON employee.employee_id = salaries.employee_id;
```

```python
df_outer = pd.merge(employees, salaries, on='employee_id', how='outer')
print(df_outer)
```

```python
df_right = pd.merge(employees, salaries, on='employee_id', how='right')
print(df_right)
```
