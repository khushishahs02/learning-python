```python
import numpy as np
import pandas as pd
```

```python
#first we will give a dictionary of input data
data = {"Name" : ['Spidey','Batey', 'Thorey', 'Captain'],
        "Age" : [22,35,100,70],
        "Salary" : [12000,23000,45000,1000000],
        "City" : ["New York","Paris", "Tokyo", "Montreal"]}
#creating a dataframe
df = pd.DataFrame(data)
print(df)
```

```python
#or we can also use lists for making a dataframe like this

data_list = [['Spidey',22, 45000, "New York"],['Batey',35,23000,'Paris'],['Captain',70,100000, 'Montreal'],["THorey",100, 45000, "Tokyo"]]
df = pd.DataFrame(data_list)
print(df)
print("\n")
#till here we havent given the column names yet
columns = ['Name', "Age", 'Salary', 'City']
df1 = pd.DataFrame(data_list, columns = columns )
print(df1)
```

```python
#selecting and indexing of columns
#df1[0] # only this will throw an error because no column is named as 0

df1['Name']

df1[['Name','Age']]
```

```python
#creating a new column
df1['Designation'] = ['engnr','doctor','mineworker', 'artist']
print(df1)
```

```python
#remove column
df1.drop('Designation', axis = 1)
#this will only remove and give a snapsot, but will not actually delete that column

# print(df1)
```

```python
#to delete a column permanently from the data frame
df1.drop('Designation', axis = 1, inplace = True)
df1

# now it is permanently deleted from the dataframe
#similary a row too can be deleted,  choosing axis = 0

df1.drop(0, axis = 0)
```

```python
#row indexing, to locate a row

print(df1.loc[0])

print(df1.loc[[0,1]])
```

```python
#selecting subsets of row and columns

df1.loc[[0,1,2]][['City', 'Salary']]
```

```python
#conditional selection
#if i want all members whose age is greater than 50

print(df1[df1['Age']>50])

print("\n")

#age less than 50 and city paris
print(df1[(df1['Age'] < 50) & (df1['City'] == "Paris")])
```
