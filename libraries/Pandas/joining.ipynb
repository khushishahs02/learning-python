```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])

print(df1)
print("\n")
print(df2)
```

```python
#merging and joining are almost the same thing, but merge is based on
#columns while join is based on index only
df1.join(df2, how = 'outer')
```

```python
df2.join(df1)
```
