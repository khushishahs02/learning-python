```python
import numpy as np
import pandas as pd
```

```python

#series is a onedimensional vertical labelled arraty. the labels can also called
#as labels. series can be created using lists, one dimensional arrays
# as well as dictionaries

labels = ['a','b','c']
mylists = [10,20,30]
myarray = np.array([10,20,30])
d = {1:10,2:20,3:30}
```

```python
#creating series

pd.Series(mylists)#this will be automatically given indices from 0 to 2
```

```python
pd.Series(mylists,index = labels)
#manually the indices as a, b, c
```

```python
pd.Series(myarray)
```

```python
pd.Series(d)
```
