```python
import numpy as np
import pandas as pd
```

```python
#concatenation is just like merging rows, just the previous indicces from dfs remain the same

df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})

print(df1)
print("\n")
print(df2)
```

```python
pd.concat([df1,df2]) #pass it as a list, and this is a concat by rows
```

```python
#if we want a concat by columns, we will mention the acis = 1
pd.concat([df1,df2], axis = 1)
```
