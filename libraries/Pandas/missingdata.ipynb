```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.isna()) # returns the true false value grid wise

#true if the value for a call in null otherwise false
```

```python
df.isna().sum()

#give columnwise sums of Nan
```

```python
df.isna().any()

#gives columnwise true if there is atleast one null value in it
```

```python
#to remove the missing data

df.dropna() # drops all the rows having null values in it
```

```python
df #the fact that drop and fill are view only and the actual data is not manipulated
```

```python
df.dropna(thresh =3 ) #drops the row if it has atleast three non nulls in it
```

```python
df
```

```python
#filling the missing data
```

```python
df.fillna(0) #fill all nan with 0s
```

```python
values = {'A':0,'B':100,"C":300,'D':400}
df.fillna(value=values)
```
