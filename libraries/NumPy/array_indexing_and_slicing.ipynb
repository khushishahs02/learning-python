```python
import numpy as np
```

```python
arr = np.arange(11,31)
print(arr)
```

```python
print(arr[1]) # to get a single element from array
```

```python
print(arr[1:5]) # slice the array from index 0 to 5 - 1 = 4
print(arr[:10]) # slice from index 0 to 9
print(arr[:]) # slice from index 0 to last element
```

```python
#numpy slicing from 2d arrays
#first we will reshape the elements
a = arr.reshape(4,5)
print(a)
```

```python
# to get a single row sliced out
row = a[1]
print(row)

```

```python
# to get another 2d vector sliced from it, first extract the row, then the column
matrix = a[2:4,1:3]
print(matrix)
```

```python
#to get a single element
a[1,2]
print(a[1,2])
```

```python
#boolean indexing
bool_index = arr % 2 == 0 # returns array of same size , but true whenever the element is even and vice versa
b =arr[bool_index] # has only the elements for which the bool_index is true i.e. the even elements
print(b)
```
