```python
import numpy as np
```

```python
#deep copy is when u slice an array and if u update it the original array doesnt get affected
a = np.array([1,2,3])
b = a.copy() # makes a deep copy
b[0] = 100
print(a)
print(b)
```

```python
#shallow copy is when tha chnages in new arrray is reflected in the original one
a = np.array([1, 2, 3, 4])
b = a.view()   # shallow copy (view)

b[0] = 100

print(a)
print(b)
```
