```python
import numpy as np
```

```python
#stacking only happens between vectors, not matrices
a = np.arange(1,5)
b = np.arange(5,9)
print(a)
print(b)
```

```python
#three types of stack

#1. horizontal stack
np.hstack((a,b))
```

```python
#vertical stack
np.vstack((a,b))
```

```python
#column stack
np.column_stack((a,b))
```
