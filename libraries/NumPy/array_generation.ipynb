```python
import numpy as np
```

```python
arr = np.arange(1,10) #range of numbers is from 1 to 9
print(arr)
```

```python
arr = np.arange(1,11,2)# range is from 1 to 10, but with step size of 2
print(arr)
```

```python
arr = np.zeros(6)
print(arr)
```

```python
arr = np.zeros((6,6))#prints 6 X 6 matrix with zeros
print(arr)
```

```python
arr = np.ones(6) # print 6 ones
print(arr)
```

```python
arr = np.linspace(1,5,10) # elements from 1 to 2, but only n equispaced elements
print(arr)
```
