```python
import numpy as np
```

```python
arr = np.random.rand(5) # prints 5 random elements from 0 to 1 , it is called normalization in statistics
print(arr)
#random is a module inbuilt in python
```

```python
arr = np.random.randn(5) # from -3 to +3, also known as standardization
print(arr)
```

```python
a = np.random.randint(6) # generates only ONE random element from 0 to 6
print(a)
```

```python
a = np.random.randint(10,20,5) # generates 5 random elements from 10 to 20
print(a)
```
